# Задача FrozenLake

FrozenLake — это классическая задача обучения с подкреплением, в которой агент должен добраться от стартовой клетки **S** до целевой клетки **G** по замёрзшему озеру, избегая провалов в проруби **H**. Поверхность льда скользкая: агент может смещаться не строго в выбранном направлении.

![FrozenLake](./frozen_lake.gif)

## Пространство действий

Действия **дискретные** (4 варианта):

* **0 — влево**
* **1 — вниз**
* **2 — вправо**
* **3 — вверх**

*(В некоторых реализациях порядок может отличаться, но смысл действий тот же.)*

## Пространство наблюдений

Наблюдение — это **номер текущей клетки** (дискретное состояние).

Если поле размера **N×N**, то количество состояний равно **N²**, а состояние кодируется числом от **0** до **N² − 1**.

Также среда задаётся картой из символов:

* **S** — старт
* **F** — безопасный лёд
* **H** — прорубь (поражение)
* **G** — цель (успех)

## Динамика среды (скольжение)

Окружение **стохастическое**: выбранное направление может быть выполнено не точно.
Типично:

* агент движется в выбранную сторону **с некоторой вероятностью**
* или “соскальзывает” в одну из перпендикулярных сторон

(Если включить режим `is_slippery=False`, переходы становятся детерминированными.)

## Система наград

Награда обычно выдаётся **только при завершении**:

* **+1**, если агент достиг клетки **G**
* **0** во всех остальных случаях (включая падение в **H** и обычные шаги)

Общая награда эпизода — сумма наград за шаги (на практике это 0 или 1).

## Начальное состояние

Эпизод начинается в клетке **S** (стартовая позиция фиксирована на карте).

## Завершение эпизода

Эпизод завершается, если:

1. агент попал в **H** (прорубь) — неудача
2. агент достиг **G** — успех
3. превышено максимальное число шагов (если задан лимит) — обычно неудача

Статус **«Принято»** ставится, если агент стабильно достигает **G** (например, высокий процент успешных эпизодов или средняя награда близка к 1 на серии запусков).
